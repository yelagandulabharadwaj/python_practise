```python
'''
Merge Two Dictionaries:
 Combine dict1 = {"a": 1, "b": 2} and dict2 = {"b": 3, "c": 4}. 
 If a key exists in both, sum their values.
'''

f1={"a": 1, "b": 2} 
f2={"b": 3, "c": 4}
res=f1.copy()
for keys,values in f2.items():
    if keys in res:
        res[keys]+=values
    else:
        res[keys]=values

print(res)

from collections import Counter
merg=dict(Counter(f1)+Counter(f2))
print(merg)


    
```

```python
'''
Given a list of integers, return all elements that appear more than once.
'''
arr=[1,1,2,3,4,4,5,6,5]
# arr=[1,1]
d=dict()
res=[]
for i in arr:
    if i in d:
        d[i]+=1
    else:
        d[i]=0
print(d)
for key,value in d.items():
    if value>=1:
        res.append(key)
print(res)
```

```python
''' 
Second Largest Number:
 Find the second largest number in a list without using the built-in sort()
'''

arr=[7,12,5,6,3,11]
sec=first=-1
for i in arr:
    if first<i :
        sec=first
        first=i
    if sec<i and first>i:
        sec=i

print(sec,first)
```

```python
class BankAccount:
    def __init__(self, balance):
        # PRIVATE variable (Hidden from the outside)
        self.__balance = balance 

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance is {self.__balance}")

    def get_balance(self):
        # Public way to "peek" at the private data
        return self.__balance

# Create an account
my_acc = BankAccount(1000)

# 1. This will FAIL (AttributeError)
# print(my_acc.__balance) 

# 2. You MUST use the methods provided
my_acc.deposit(500)
print(f"Current Balance: {my_acc.get_balance()}")
```

```python
from abc import ABC, abstractmethod

class Greet(ABC):
    @abstractmethod
    def say_hello(self):
        pass  # Abstract method

class English(Greet):
    def say_hello(self):
        return "Hello!"

g = English()
print(g.say_hello())
```

```python


def dec(func):
    print('entered dec function')
    def wrapper(name,id,dept):
        print('before main')
        func(name,id,dept)
        print('after main')
    return wrapper

@dec
def my_details(name,id,dept):
    print(name,'--',id,'--',dept)


my_details('bharat',123,'IT')
```

```python
''' 
multiple decorators
'''
def dec1(func):
    def wrap(name,id):
        print('before dec1')
        func(name,id)
        print('after dec1')
    return wrap

def dec2(func):
    def wrapper(name,id):
        print('before dec2')
        func(name,id)  #dec1(name,id)
        print('after dec2')
    return wrapper


# it works like dec1(dec2(greet))
@dec1
@dec2
def greet(name,id):
    print(name,id)

greet('bharadwaj',12)
```
