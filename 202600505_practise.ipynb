```python
'''
MAJORITY ELEMENT

The majority element is the element that appears more than ⌊n / 2⌋ times. You may assume that the majority element always exists in the array.

Input: nums = [3,2,3]
Output: 3

Input: nums = [2,2,1,1,1,2,2]
Output: 2

'''
nums = [2,2,1,1,1,2,2]
tre=len(nums)//2
hmap={}
for i in nums:
    if i in hmap:
        hmap[i]+=1
    else:
        hmap[i]=1
print(hmap,tre)
c=0
for i in hmap:
    if hmap[i]>tre:
        c=1
        print('found it:',i, hmap[i])
if c==0:
    print('no ele found')
```

```python
'''
Collect rain water
You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the ith line are (i, 0) and (i, height[i]).
Find two lines that together with the x-axis form a container, such that the container contains the most water.
Return the maximum amount of water a container can store.

Input: height = [1,8,6,2,5,4,8,3,7]
Output: 49

'''

'''
core logic : water[i]= min( leftmax,right_max )-height
sum of water array
'''

wtr=[1,8,6,2,5,4,8,3,7]

i,j=0,len(wtr)-1
max_water=0
while i<=j:
    h=min(wtr[i],wtr[j])
    width=j-i
    area=h*width

    max_water=max(max_water,area)

    if wtr[i]<wtr[j]:
        i+=1
    else:
        j-=1

print(max_water)

```

```python
'''
K Closest Points to Origin
Given an array of points where points[i] = [xi, yi] represents a point on the X-Y plane and an integer k, return the k closest points to the origin (0, 0).
The distance between two points on the X-Y plane is the Euclidean distance (i.e., √(x1 - x2)2 + (y1 - y2)2).

Input: points = [[1,3],[-2,2]], k = 1
Output: [[-2,2]]
Explanation:
The distance between (1, 3) and the origin is sqrt(10).
The distance between (-2, 2) and the origin is sqrt(8).
Since sqrt(8) < sqrt(10), (-2, 2) is closer to the origin.
We only want the closest k = 1 points from the origin, so the answer is just [[-2,2]].

Input: points = [[3,3],[5,-1],[-2,4]], k = 2
Output: [[3,3],[-2,4]]
Explanation: The answer [[-2,4],[3,3]] would also be accepted.
 
'''
import math
grid= [[3,3],[5,-1],[-2,4]]# [[1,3],[-2,2]]
k=2#1
hmap={}
for val in grid:
    x,y=val[0],val[1]
    hmap[x,y]= math.sqrt( (x**2) + (y**2))

hmap_srted= sorted(hmap.items(), key=lambda x:x[-1], reverse=False)
res = []
for val in hmap_srted:
    if k == 0:
        break
    res.append(val)
    k -= 1
print(res)
```
