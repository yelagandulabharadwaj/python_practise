```python
'''
Product of Array Except Self
expected: Time complexity goal: O(n)
Problem description
Given an integer array nums, the goal is to return an array answer where answer[i] equals 
the product of all elements in nums except nums[i]. 
Division is not allowed, so the solution must compute cumulative products efficiently.
 For instance, with input [1,2,3,4], the expected output is [24,12,8,6].
'''

arr=[1,2,3,4]
res=[1]*len(arr)
for i in range(len(arr)):
    for j in range(len(arr)):
        if i!=j:
            res[i]*=arr[j]
print(res)
'''
expected: Time complexity goal: O(n)
'''
arr=[-1,1,0,-3,3]
all_mul=1
res2=[1]*len(arr)
for i in range(len(arr)):
    all_mul*=arr[i]
print(all_mul)
if 0 not in arr:
    for i in range(len(arr)):
        res2[i]=all_mul//arr[i]

    print(res2)
else:
    ind=arr.index(0)
    res3=[1]*len(arr)
    all_mul=1
    for i in range(len(arr)):
        if i!=ind:
            all_mul*=arr[i]
    for i in range(len(arr)):
        if i!=ind:
            res3[i]=0
        else:
            res3[i]=all_mul
    print(res3,'\n\n')

'''
if required without division

below is the optimal solution O(n) without division
'''

# arr=[1,2,3,4]
arr=[3,4,6,1,2]
leftie=1
res1=[1]*len(arr)
for i in range(len(arr)):
    res1[i]=leftie
    leftie*=arr[i]

print(res1)


rightie=1
res2=[1]*len(arr)
for i in range(len(arr)-1,-1,-1):
    res2[i]=rightie
    rightie*=arr[i]
print(res2)

final_res=[1]*len(arr)
for  i  in range(len(arr)):
    final_res[i]=res1[i]*res2[i]
print(final_res)
```

```python
'''
Longest common subsequence

ex: [100, 4, 200, 1, 3, 2]
res: 4 (those are 1,2,3,4)
consecutive (x,x+1,x+2,...) irrespective of order in array
'''
arr=[100, 4, 200, 1, 3, 2]
res=[0]*len(arr)
for i in range(len(arr)):
    m=arr[i]
    c=0
    while m in arr:
        c+=1
        res[i]=c
        m=m+1
print(res)
'''
its actually O(n^^3) implementation above
for loop (n)
while loop (n)
m in arr (n)
in total O(n^^3)
'''

arr=[100, 4, 200, 1, 3, 2]
opt_set=set(arr)
res=[0]*len(arr)
for i in range(len(arr)):
    m=arr[i]
    c=0
    while m in opt_set:
        c+=1
        res[i]=c
        m=m+1
print(res)
print(opt_set)
'''
above 
for loop  -- O(n)
while loop run only restrcited period of time as it is executing in set , so best case O(1)  worst O(n)
m in set() it takes only O(1) 
'''
```
